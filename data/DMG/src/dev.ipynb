```python
def compact(series, keep_repetitions=False, as_string=True, sep="&semi;"):
    u = series.unique()
    if len(u) == 0:
        return "" if as_string else None
    elif len(u) == 1:
        return series.iloc[0]
    else:
        x = series if keep_repetitions else u
        return sep.join(x) if as_string else list(x)


# def implode(df, on="object_number", as_string=True, sep="&semi;"):
def _implode(subdf):
    return np.asarray([compact(subdf[col].dropna(), as_string=as_string, sep=sep) for col in subdf.columns])# + [subdf.name]
on="object_number"
as_string=True
sep="&semi;"

small = df.iloc[:10]
rows = small.groupby(on, as_index=False).progress_apply(_implode)
x = pd.DataFrame(rows.tolist())#.reset_index()
x.columns = df.columns

# implode(df.iloc[:10])




# df.columns

# small.groupby(on, as_index=True).progress_apply(lambda sub: pd.Series([sub[c] for c in sub.columns]))


# pd.Series([small.iloc[:10][c].iloc[0] for c in small.columns])

x.shape
```
